```python
class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class Restaurant:
    def __init__(self):
        # Fixed restaurant & owner details
        self.name = "Food Plaza"
        self.owner = "Kushal Prajapati"
        self.menu = []

    def add_item(self, item):
        self.menu.append(item)

    def show_menu(self):
        print("\n------ MENU ------")
        for i, item in enumerate(self.menu, start=1):
            print(f"{i}. {item.name} - ₹{item.price}")


class Customer:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone   
        self.order = []

    def add_order(self, item, qty):
        self.order.append((item, qty))
        print(f"{item.name} x{qty} added ✅")

    def generate_bill(self, restaurant):
        print("\n=========== BILL ===========")
        print(f"Restaurant : {restaurant.name}")
        print(f"Owner      : {restaurant.owner}")
        print("----------------------------")
        print(f"Customer   : {self.name}")
        print(f"Phone      : {self.phone}")
        print("----------------------------")

        total = 0
        print("\nItems Ordered:")
        for item, qty in self.order:
            price = item.price * qty
            total += price
            print(f"{item.name} x{qty} = ₹{price}")

        print("----------------------------")
        gst = total * 0.05
        final_total = total + gst

        print(f"Subtotal : ₹{total}")
        print(f"GST (5%) : ₹{gst}")
        print(f"Total    : ₹{final_total}")
        print("=========== THANK YOU ===========")


# Main Program
def main():
    restaurant = Restaurant()

    # Menu items (10+)
    items = [
        MenuItem("Pizza", 250),
        MenuItem("Burger", 120),
        MenuItem("Pasta", 180),
        MenuItem("Sandwich", 100),
        MenuItem("Coffee", 80),
        MenuItem("Tea", 50),
        MenuItem("Cold Drink", 90),
        MenuItem("Noodles", 150),
        MenuItem("Manchurian", 160),
        MenuItem("Fries", 110),
        MenuItem("Ice Cream", 130)
    ]

    for item in items:
        restaurant.add_item(item)

    # Customer Input
    print("Enter Customer Details:")
    name = input("Name: ")
    phone = input("Phone: ")

    customer = Customer(name, phone)

    # Ordering
    while True:
        restaurant.show_menu()
        choice = int(input("Select item number (0 to finish): "))

        if choice == 0:
            break

        if 1 <= choice <= len(restaurant.menu):
            qty = int(input("Enter quantity: "))
            customer.add_order(restaurant.menu[choice - 1], qty)
        else:
            print("Invalid choice ❌")

    # Bill
    customer.generate_bill(restaurant)


# Run
main()
```
